### Using `gTTS` (Google Text-to-Speech)

`gTTS` is a Python library and CLI tool to interface with Google Translate's text-to-speech API. It's easy to use and provides natural-sounding voices.

First, you need to install it:

```python
!pip install gTTS
```

Now, here's how you can use `gTTS` to convert text to speech and even save it to an MP3 file:

```python
from gtts import gTTS
import os

# The text you want to convert to speech
text = "Hello Evryone, welcome to your text to voice project with gTTS!"

# Choose language (e.g., 'en' for English)
language = 'en'

# Pass the text and language to the gTTS object
# You can also specify slow=True for slower speech
mytobj = gTTS(text=text, lang=language, slow=False)

# Save the converted audio to an mp3 file
file_name = "welcome_message.mp3"
mytobj.save(file_name)

print(f"Audio saved as {file_name}")

# To play the audio in Colab, you can use IPython.display.Audio
from IPython.display import Audio
Audio(file_name)
```

### Using `ipywidgets` for an interactive Text-to-Speech interface

`ipywidgets` provides interactive HTML widgets for Jupyter notebooks. This allows you to create user interfaces that run directly within your Colab output, without needing a separate display server. It's perfect for creating simple interactive tools.

```python
from gtts import gTTS
from IPython.display import Audio, display
import ipywidgets as widgets
from IPython.display import HTML

def text_to_speech(text_input):
    if not text_input:
        print("Please enter some text.")
        return

    # Generate speech using gTTS
    tts = gTTS(text=text_input, lang='en', slow=False)
    audio_file = "output.mp3"
    tts.save(audio_file)

    print("Generating audio...")
    display(Audio(audio_file, autoplay=True)) # Autoplay the audio

# Create a text area widget
text_area = widgets.Textarea(
    value="Hello everyone, my name is Shweta and this is my Text-to-Speech multi-voice AI app.In this project, I have built an application where the user can type any text and convert it into human-sounding speech.It supports multiple voices, male/female, and different languages using Microsoft Edge Neural voices.",
    placeholder='Type your text here',
    description='Text:',
    disabled=False
)

# Create a button widget
button = widgets.Button(
    description='Convert to Speech',
    disabled=False,
    button_style='success', # 'success', 'info', 'warning', 'danger' or ''
    tooltip='Click to convert text to speech',
    icon='microphone' # (FontAwesome names without the fa- prefix)
)

# Define the action for the button click
button.on_click(lambda b: text_to_speech(text_area.value))

# Display the widgets
display(text_area, button)

# Update the lang_select with more languages if the user requests it
# For example, in a separate cell or by modifying this one directly:
# Note: This part assumes the existing widgets were created in this cell.
# To modify an existing widget, you would generally interact with its instance.
# For this demonstration, I'm showing how to add more options assuming
# a re-execution of the widget creation.
# If the user wants to *add* dynamically, more complex logic would be needed.

# Re-create language selector with more options
# This assumes a fresh run of the cell, or that lang_select can be modified directly
# If this were an interactive session, you might need to find the existing widget and update it.
lang_select = widgets.Dropdown(
    options=['en', 'es', 'fr', 'de', 'it', 'pt', 'ja', 'ko', 'zh-CN', 'hi', 'bn', 'gu', 'kn', 'ml', 'mr', 'pa', 'ta', 'te', 'ur'],
    value='en',
    description='Language:',
    disabled=False,
)

# Re-display the updated dropdown (if in an interactive context)
display(lang_select)
```

```python
# 3️⃣ Create Streamlit app
%%writefile app.py
import streamlit as st
from gtts import gTTS
import os

st.title("🎤 Text to Voice App")

text = st.text_area("Enter your text:")

lang = st.selectbox("Select Language", ["en", "hi", "es", "fr", "de"])
voice_type = st.selectbox("Voice Type", ["Female", "Male"])

if st.button("Generate Audio"):
    if text.strip() != "":
        tld = "com" if voice_type == "Female" else "co.uk"
        tts = gTTS(text=text, lang=lang, tld=tld)
        tts.save("output.mp3")
        st.audio("output.mp3", format="audio/mp3")
        st.success("Audio Generated! You can play or download it.")
    else:
        st.warning("Please enter some text!")

```
